# Bar Plot — When to Use It

A bar plot compares a numeric value across discrete categories using rectangular bars of varying height, making rank differences easy to spot.

## Real-World Use Cases

**1. Comparing categories** — Compare the happiness scores of the top countries in the dataset to see which rank highest or lowest at a glance.

**2. Sales by product** — Compare revenue or units sold per product so the best- and worst-performing items are immediately visible.

**3. Population by city** — Compare population sizes across cities to identify the largest and smallest in a single view.
