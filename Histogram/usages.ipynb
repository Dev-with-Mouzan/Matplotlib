# Histogram — When to Use It

A histogram groups a continuous variable into bins and counts how many observations fall into each, revealing the shape of a distribution.

## Real-World Use Cases

**1. Frequency of happiness scores** — See how happiness scores are distributed across countries and where most of them cluster.

**2. Age distribution** — Understand how a population or customer base is spread across age groups.

**3. Order amounts** — Inspect how order sizes are distributed to spot typical spending and unusual outliers.
