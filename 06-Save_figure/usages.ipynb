# Save Figure — When to Use It

Saving a figure exports a plot as a file so it can be used in reports, presentations, or publication-ready documents.

## Real-World Use Cases

**1. Report or publication** — Save the plot as PNG or PDF at high DPI (e.g., 300) for crisp print quality.

**2. Web or presentation** — Export as SVG for sharp vector graphics, or as PNG with a transparent background.

**3. Reproducibility** — Automatically save generated plots so results can be reused without re-running the notebook.
