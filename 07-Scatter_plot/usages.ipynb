# Scatter Plot — When to Use It

A scatter plot shows the relationship between two numeric variables as individual points, making correlations visible.

## Real-World Use Cases

**1. GDP vs Happiness** — Check whether richer countries tend to report higher happiness scores in the dataset.

**2. Study hours vs exam score** — Spot how study time relates to exam performance across students.

**3. Advertising spend vs sales** — Detect whether more marketing spend tends to increase sales.
