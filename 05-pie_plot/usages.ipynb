# Pie Chart — When to Use It

A pie chart shows parts of a whole as slices proportional to their share, best used with few categories.

## Real-World Use Cases

**1. Share of top countries** — Show how much each of the top 5 countries contributes to total happiness.

**2. Market share** — Display each company's share of the total market.

**3. Budget allocation** — Break a budget into categories so the largest spending area is obvious.
