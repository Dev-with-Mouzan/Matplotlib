# Step Plot — When to Use It

A step plot displays data that changes in discrete jumps rather than continuously, matching how many real-world values actually change.

## Real-World Use Cases

**1. Rank transitions** — Show happiness values that change only at specific positions across the top countries.

**2. Price changes** — Display a product price that stays constant until the moment it is updated.

**3. CPU or memory usage** — Visualize resource usage that stays flat between sampling moments.
