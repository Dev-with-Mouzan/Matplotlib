# Fill Between Plot — When to Use It

Fill Between shades the area between a line and a baseline (or between two lines), ideal for highlighting regions of interest.

## Real-World Use Cases

**1. Confidence intervals** — Shade the band between the upper and lower whisker values around happiness scores to emphasize uncertainty.

**2. Target range** — Highlight a target band (e.g., an acceptable temperature range) on a time-series line chart.

**3. Min–max envelope** — Shade the range between minimum and maximum values across categories to show variability.
