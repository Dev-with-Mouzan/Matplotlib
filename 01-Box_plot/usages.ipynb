# Box Plot — When to Use It

A box plot summarizes a distribution through quartiles, whiskers and outliers, so you can quickly judge spread and unusual values.

## Real-World Use Cases

**1. Distribution of happiness scores** — Show the median, spread and any outlier countries in the Happiness column of the dataset.

**2. Salary by department** — Compare pay ranges across departments to detect compensation differences and outliers.

**3. Test scores by class** — Compare medians and the spread of scores to see which class performs more consistently.
