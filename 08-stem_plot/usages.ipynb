# Stem Plot — When to Use It

A stem plot marks discrete data points with markers on vertical lines, common in signal processing and other discrete-value fields.

## Real-World Use Cases

**1. Discrete samples** — Plot individual happiness values per country as discrete markers above a baseline.

**2. Event counts** — Show the number of events recorded per day across a week.

**3. Impulse or signal data** — Visualize discrete signal samples in engineering or scientific work.
